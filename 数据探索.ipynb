```python
import pandas as pd
import numpy as np
import os


alpha = 0.5
beta = 0.5
W_0 = 1  
gamma = 1  

industry_etf_map = {
    "化工": "159870.OF.csv",
    "传统能源": "159930.OF.csv",
    "金融": "510230.OF.csv",
    "医药": "512010.OF.csv",
    "消费": "159928.OF.csv",
    "房地产": "512200.OF.csv",
    "电动车新能源": "515030.OF.csv",
    "高端装备": "516320.OF.csv",
    "旅游": "159766.OF.csv",
    "半导体与信息技术": "159939.OF.csv"
}


base_data_path = r"C:\Users\wrz\Desktop\FYPProject\Data"
policy_sentiment_path = r"C:\Users\wrz\Desktop\FYPProject\adjusted_policy_sentiment_score_M.csv"
market_sentiment_path = r"C:\Users\wrz\Desktop\FYPProject\market_sentiment_factor_M.csv"


policy_df = pd.read_csv(policy_sentiment_path)
market_df = pd.read_csv(market_sentiment_path)

combined = policy_df.merge(market_df, on="period", how="left")


def calculate_weights(group, alpha=alpha, beta=beta, gamma=gamma):
    group["W_prime"] = W_0 * (1 + alpha * group["S_adjusted"] + beta * group["M"])
    exp_vals = np.exp(gamma * group["W_prime"])
    group["W_final"] = exp_vals / exp_vals.sum() * 100  
    return group


weight_result = combined.groupby("period", group_keys=False).apply(calculate_weights)


output_path = r"C:\Users\wrz\Desktop\FYPProject\W_final_softmax.csv"
weight_result.to_csv(output_path, index=False)

print(" 计算完成，保存至：", output_path)
```
